# EntrySignalLab quick start
Install the local package before running. This synthetic example reports conditional entry candidates and separately compares a zero non-entry reference.

```python
from entrysignallab import ModelParams, DecisionBounds, compare
p = ModelParams(announcement_cost=0.0002, deployment_cost=0.012, learning_rate=12)
report = compare(p, DecisionBounds(announcement_upper=2), rival_delay=0.15)
[(r["strategy"], r["delay"], r["announcement"], r["payoff"]) for r in report["results"]]
```

```python
report["co_best"], report["preferred_with_outside_option"]
```
